```python
# Travelling Salesman Problem

from itertools import permutations

# Number of cities
cities = 4

# Cost matrix
graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

# Starting city
start = 0

# Store minimum cost and best route
minimum_cost = float('inf')
best_route = []

# Generate all possible routes
other_cities = [i for i in range(cities) if i != start]

for route in permutations(other_cities):

    current_cost = 0
    current_city = start

    # Calculate the route cost
    for next_city in route:
        current_cost += graph[current_city][next_city]
        current_city = next_city

    # Return to starting city
    current_cost += graph[current_city][start]

    # Update minimum cost
    if current_cost < minimum_cost:
        minimum_cost = current_cost
        best_route = [start] + list(route) + [start]

# Display the result
print("Best Route:")

for city in best_route:
    print(city, end=" ")

print("\nMinimum Travel Cost:", minimum_cost)
```
