```python
## import time

# Bubble Sort
def bubble_sort(arr):
    a = arr.copy()

    for i in range(len(a) - 1):
        swapped = False

        for j in range(len(a) - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        if not swapped:
            break

    return a


# Selection Sort
def selection_sort(arr):
    a = arr.copy()

    for i in range(len(a) - 1):
        min_index = i

        for j in range(i + 1, len(a)):
            if a[j] < a[min_index]:
                min_index = j

        a[i], a[min_index] = a[min_index], a[i]

    return a


# Insertion Sort
def insertion_sort(arr):
    a = arr.copy()

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j = j - 1

        a[j + 1] = key

    return a


# Merge Sort
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


# Quick Sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[-1]

    left = []
    right = []

    for i in range(len(arr) - 1):
        if arr[i] <= pivot:
            left.append(arr[i])
        else:
            right.append(arr[i])

    return quick_sort(left) + [pivot] + quick_sort(right)


# Input
n = int(input("Enter number of elements: "))

arr = []

print("Enter the elements:")

for i in range(n):
    arr.append(int(input()))


# Display original array
print("\nOriginal Array:")
print(arr)


# Bubble Sort and time
start = time.perf_counter()
bubble_result = bubble_sort(arr)
end = time.perf_counter()
bubble_time = (end - start) * 1000000


# Selection Sort and time
start = time.perf_counter()
selection_result = selection_sort(arr)
end = time.perf_counter()
selection_time = (end - start) * 1000000


# Insertion Sort and time
start = time.perf_counter()
insertion_result = insertion_sort(arr)
end = time.perf_counter()
insertion_time = (end - start) * 1000000


# Merge Sort and time
start = time.perf_counter()
merge_result = merge_sort(arr)
end = time.perf_counter()
merge_time = (end - start) * 1000000


# Quick Sort and time
start = time.perf_counter()
quick_result = quick_sort(arr)
end = time.perf_counter()
quick_time = (end - start) * 1000000


# Display results
print("\nSorted Arrays:")

print("Bubble Sort    :", bubble_result)
print("Selection Sort :", selection_result)
print("Insertion Sort :", insertion_result)
print("Merge Sort     :", merge_result)
print("Quick Sort     :", quick_result)


# Display execution time
print("\nExecution Time:")

print("Bubble Sort    :", round(bubble_time, 2), "microseconds")
print("Selection Sort :", round(selection_time, 2), "microseconds")
print("Insertion Sort :", round(insertion_time, 2), "microseconds")
print("Merge Sort     :", round(merge_time, 2), "microseconds")
print("Quick Sort     :", round(quick_time, 2), "microseconds")


# Time complexity
print("\nTime Complexity:")

print("Algorithm       Best Case     Average Case     Worst Case")
print("Bubble Sort     O(n)          O(n^2)           O(n^2)")
print("Selection Sort  O(n^2)        O(n^2)           O(n^2)")
print("Insertion Sort  O(n)          O(n^2)           O(n^2)")
print("Merge Sort      O(n log n)    O(n log n)       O(n log n)")
print("Quick Sort      O(n log n)    O(n log n)       O(n^2)")
```
