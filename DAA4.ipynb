```python
import time

# Input from user
n = int(input("Enter a number: "))

# Iterative Method
start_time = time.perf_counter()

fact_iterative = 1
for i in range(1, n + 1):
    fact_iterative = fact_iterative * i

end_time = time.perf_counter()

iterative_time = (end_time - start_time) * 1_000_000


# Recursive Method
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)


start_time = time.perf_counter()

fact_recursive = factorial_recursive(n)

end_time = time.perf_counter()

recursive_time = (end_time - start_time) * 1_000_000


# Display results
print("\nFactorial using Iterative Method:", fact_iterative)
print("Execution Time:", round(iterative_time, 2), "microseconds")

print("\nFactorial using Recursive Method:", fact_recursive)
print("Execution Time:", round(recursive_time, 2), "microseconds")

# Time Complexity
print("\nTime Complexity:")
print("Iterative Method : O(n)")
print("Recursive Method : O(n)")
```
