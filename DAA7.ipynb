```python
import time

# Making Change Problem using Dynamic Programming

print("========================================")
print("       MAKING CHANGE PROBLEM")
print("       USING DYNAMIC PROGRAMMING")
print("========================================")

# Get number of coins
n = int(input("\nEnter number of coin types: "))

coins = []

# Get coin values
print("\nEnter the coin values:")
for i in range(n):
    coin = int(input(f"Enter coin {i + 1}: "))
    coins.append(coin)

# Get amount
amount = int(input("\nEnter the amount: "))

# Start execution time
start_time = time.perf_counter()

# Create DP array
dp = [amount + 1] * (amount + 1)

# Store the last coin used
used_coin = [-1] * (amount + 1)

# Base case
dp[0] = 0

# Dynamic Programming
for i in range(1, amount + 1):

    for coin in coins:

        if coin <= i:

            if dp[i - coin] + 1 < dp[i]:

                dp[i] = dp[i - coin] + 1
                used_coin[i] = coin

# End execution time
end_time = time.perf_counter()

# Calculate execution time
execution_time = (end_time - start_time) * 1000000

# Check whether change is possible
if dp[amount] == amount + 1:

    print("\nChange cannot be made for the given amount.")

else:

    print("\nMinimum number of coins:", dp[amount])

    # Find selected coins
    selected_coins = []
    current_amount = amount

    while current_amount > 0:

        coin = used_coin[current_amount]

        if coin == -1:
            break

        selected_coins.append(coin)
        current_amount = current_amount - coin

    print("Coins used:", selected_coins)

# Display DP table
print("\nDP Table:")
print("Amount\tMinimum Coins")

for i in range(amount + 1):
    if dp[i] == amount + 1:
        print(i, "\tNot Possible")
    else:
        print(i, "\t", dp[i])

# Display execution time
print("\nExecution Time:",
      round(execution_time, 2),
      "microseconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case    : O(n × amount)")
print("Average Case : O(n × amount)")
print("Worst Case   : O(n × amount)")

# Display space complexity
print("\nSpace Complexity: O(amount)")
```
