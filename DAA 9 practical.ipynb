```python
# Prim's Algorithm to find Minimum Spanning Tree

# Number of vertices
vertices = 5

# Graph represented using an adjacency matrix
graph = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]

# To check whether a vertex is visited or not
visited = [False] * vertices

# Start from the first vertex
visited[0] = True

print("Minimum Spanning Tree:")
print("Edge\tWeight")

# Repeat until all vertices are connected
for i in range(vertices - 1):

    minimum = 999999
    start = 0
    end = 0

    # Find the minimum weight edge
    for j in range(vertices):
        if visited[j]:

            for k in range(vertices):
                if not visited[k] and graph[j][k] != 0:

                    if graph[j][k] < minimum:
                        minimum = graph[j][k]
                        start = j
                        end = k

    # Display the selected edge
    print(start, "-", end, "\t", minimum)

    # Mark the new vertex as visited
    visited[end] = True

print("Minimum Spanning Tree completed.")
```
